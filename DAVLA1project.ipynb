```python
import pandas as pd
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import nltk

# Download VADER lexicon (run once)
nltk.download('vader_lexicon')

# Load the raw dataset
df = pd.read_csv('sentimentdataset.csv')

# Automatically detect the text column based on common naming conventions
possible_text_columns = ['text', 'Text', 'Comment', 'Message', 'Tweet', 'Content', 'Body']
text_col = next((col for col in possible_text_columns if col in df.columns), df.columns[1])

# Drop rows with missing text messages
df = df.dropna(subset=[text_col])

# Initialize VADER sentiment analyzer
sia = SentimentIntensityAnalyzer()

# Compute sentiment compound scores and assign categorical labels
df['Sentiment_Score'] = df[text_col].apply(lambda x: sia.polarity_scores(str(x))['compound'])
df['Sentiment_Label'] = df['Sentiment_Score'].apply(
    lambda x: 'Positive' if x > 0.05 else ('Negative' if x < -0.05 else 'Neutral')
)

# Export polished CSV for your dashboard
df.to_csv('cleaned_sentiment_data.csv', index=False)
print("Data processing complete! 'cleaned_sentiment_data.csv' is ready.")
```
